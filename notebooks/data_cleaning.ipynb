```python
# clean the column names
def clean_columnn_name(col):
    col = str(col)
    col = col.strip()
    col = col.lower()
    col = col.replace(" ","_")
    col = col.replace("-","_")
    col = col.replace("/","_")
    col = col.replace("__","_")
    return col
```
