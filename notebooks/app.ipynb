````python
import shutil
import os

project_name = "uganda-food-price-prediction"

# Raw dataset
if os.path.exists("wfp_food_prices_uga.csv"):
    shutil.copy("wfp_food_prices_uga.csv", f"{project_name}/data/raw/wfp_food_prices_uga.csv")

# Cleaned ML-ready dataset
if os.path.exists("wfp_uganda_food_prices_ml_ready.csv"):
    shutil.copy("wfp_uganda_food_prices_ml_ready.csv", f"{project_name}/data/processed/wfp_uganda_food_prices_ml_ready.csv")

# Final model
if os.path.exists("random_forest_food_price_model.pkl"):
    shutil.copy("random_forest_food_price_model.pkl", f"{project_name}/models/random_forest_food_price_model.pkl")

# App reference dataset
if os.path.exists("food_price_input_reference.csv"):
    shutil.copy("food_price_input_reference.csv", f"{project_name}/app/food_price_input_reference.csv")

# App file
if os.path.exists("app.py"):
    shutil.copy("app.py", f"{project_name}/app/app.py")

# Predictions
if os.path.exists("final_food_price_predictions.csv"):
    shutil.copy("final_food_price_predictions.csv", f"{project_name}/data/processed/final_food_price_predictions.csv")

print("Available files copied.")
````

````python
readme_text = """
# Uganda Food Price Prediction

## Project Overview

This project uses World Food Programme Uganda market price data to predict retail food prices in Ugandan markets. The machine learning task is a regression problem where the target variable is the next month retail price of a food commodity.

## Dataset

The dataset contains Uganda market price records with information on:

- Date
- Region and district
- Market
- Commodity
- Commodity category
- Unit
- Retail price
- USD price
- Latitude and longitude

The raw dataset contains over 32,000 records. After filtering to retail food prices and creating lag features, the final ML-ready dataset contains over 20,000 records.

## Machine Learning Goal

The goal is to predict the next month retail price of a food commodity in a Ugandan market.

Target variable:

`next_month_price`

## Features Used

The model uses:

- Year
- Month
- Quarter
- Region
- District
- Market
- Latitude
- Longitude
- Category
- Commodity
- Unit
- Current price
- Previous price
- Price change
- Percentage price change
- 3-month average price
- 6-month average price

## Models Used

Three models were compared:

1. Baseline model
2. Random Forest Regressor
3. Neural Network Regressor

## Model Performance

| Model | MAE | RMSE | R2 |
|---|---:|---:|---:|
| Baseline | 557.70 | 1975.42 | 0.8586 |
| Random Forest | 533.20 | 1697.70 | 0.8955 |
| Neural Network | 691.14 | 1760.43 | 0.8877 |

The Random Forest Regressor was selected as the final model because it achieved the best overall performance.

## Demo Interface

A Gradio demo interface was created to allow users to select a commodity and market, enter price values, and predict the next month retail food price.

## Project Structure

```text
uganda-food-price-prediction/
│
├── data/
│   ├── raw/
│   └── processed/
│
├── notebooks/
│
├── models/
│
├── reports/
│   └── figures/
│
├── app/
│
├── presentation/
│
├── README.md
├── requirements.txt
└── .gitignore """
````
